## 3. Главная страница

```python
rows = []

for product in products:
    rows.append(
        f"""
        <tr>
            <td>
                <a href="{product['file']}" class="product-link">
                    {product['name']}
                </a>
            </td>
            <td>{product['price']:,} ₽</td>
            <td>{product['stock']} м³</td>
            <td class="{product['status_class']}">
                {product['status']}
            </td>
        </tr>
        """
    )

index_html = f"""
<!DOCTYPE html>
<html lang="ru">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Склад строительных материалов</title>
    <link rel="stylesheet" href="style.css">
</head>
<body>
    <div class="container">
        <div class="header">
            <h1>СКЛАД СТРОИТЕЛЬНЫХ МАТЕРИАЛОВ</h1>
            <p>Профессиональные материалы для строительства и ремонта</p>
        </div>

        <div class="table-container">
            <h2>ТАБЛИЦА ТОВАРОВ НА СКЛАДЕ</h2>

            <table>
                <thead>
                    <tr>
                        <th>Название товара</th>
                        <th>Стоимость за 1 м³</th>
                        <th>Остаток</th>
                        <th>Статус наличия</th>
                    </tr>
                </thead>
                <tbody>
                    {''.join(rows)}
                </tbody>
            </table>
        </div>

        <div class="legend">
            <h3>ЛЕГЕНДА СТАТУСОВ ЗАПАСА</h3>
            <p class="status-high">Высокий запас: больше 100 м³</p>
            <p class="status-medium">Средний запас: 50-100 м³</p>
            <p class="status-low">Низкий запас: меньше 50 м³</p>
        </div>
    </div>
</body>
</html>
"""
```

## 4. Страницы товаров

```python
def make_product_page(index):
    product = products[index]
    next_file = "index.html" if index == len(products) - 1 else products[index + 1]["file"]

    details_html = "".join(
        f"<p><strong>{title}:</strong> {value}</p>"
        for title, value in product["details"]
    )

    return f"""
<!DOCTYPE html>
<html lang="ru">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>{product['name']}</title>
    <link rel="stylesheet" href="style.css">
</head>
<body>
    <div class="container">
        <div class="header">
            <h1>{product['name'].upper()}</h1>
            <p>Подробная информация о товаре</p>
        </div>

        <div class="product-info">
            <h2>ХАРАКТЕРИСТИКИ ТОВАРА</h2>
            <p><strong>Название товара:</strong> {product['full_name']}</p>
            <p><strong>Стоимость за 1 м³:</strong> {product['price']:,} ₽</p>
            <p><strong>Остаток на складе:</strong> {product['stock']} м³</p>
            <p class="{product['status_class']}">
                <strong>Статус наличия:</strong> {product['status']}
            </p>
            {details_html}
        </div>

        <img
            src="{product['image']}"
            alt="{product['name']}"
            class="product-image"
        >

        <div class="product-info">
            <h2>ОПИСАНИЕ И ПРИМЕНЕНИЕ</h2>
            <p>{product['description']}</p>
        </div>

        <div class="navigation">
            <a href="index.html" class="nav-button">На главную</a>
            <a href="{next_file}" class="nav-button">Следующая страница →</a>
        </div>
    </div>
</body>
</html>
"""
```

## 5. Создание файлов сайта

```python
site_dir = Path("warehouse_site")
images_dir = site_dir / "images"

site_dir.mkdir(exist_ok=True)
images_dir.mkdir(exist_ok=True)

(site_dir / "style.css").write_text(
    style_css,
    encoding="utf-8"
)

(site_dir / "index.html").write_text(
    index_html,
    encoding="utf-8"
)

for i, product in enumerate(products):
    (site_dir / product["file"]).write_text(
        make_product_page(i),
        encoding="utf-8"
    )

print("Созданные файлы:")
for path in sorted(site_dir.rglob("*")):
    print(path)
```
